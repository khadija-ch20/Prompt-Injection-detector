```python
from datasets import load_dataset
import pandas as pd

# Charger le dataset
ds = load_dataset("cyberec/Prompt-injection-dataset", "core")

# Convertir en DataFrame
df_train = ds["train"].to_pandas()
df_val = ds["validation"].to_pandas()
df_test = ds["test"].to_pandas()

# Vérifier  la liste des noms de colonnes de DataF
print("**************************************************************************************************")
print(df_train.columns)

# Sauvegarder en CSV
df_train.to_csv("../data/train.csv", index=False)  #index =False = ne pas sauvgrader la colonne d'index.
df_val.to_csv("../data/val.csv", index=False)
df_test.to_csv("../data/test.csv", index=False)

# Vérifier tailles
print("*************************************************************************************************")
print("Train : ",len(df_train)," \nValidation : " ,len(df_val),  "\nTest : ",len(df_test))
```
